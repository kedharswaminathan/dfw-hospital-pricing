```python
df = pd.DataFrame({
    "hospital": ["Baylor Scott & White", "Texas Health", "Methodist", "Medical City", "HCA North TX"],
    "procedure": ["Knee MRI"] * 5,
    "negotiated_rate": [420, 780, 510, 1150, 690],
})
df
```

```python
fig, ax = plt.subplots(figsize=(8, 4))
ax.bar(df["hospital"], df["negotiated_rate"], color="steelblue")
ax.set_title("Sanity check: fake DFW knee MRI rates")
ax.set_ylabel("Negotiated rate ($)")
ax.set_xticklabels(df["hospital"], rotation=20, ha="right")
plt.tight_layout()
plt.show()
```
